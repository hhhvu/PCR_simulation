## Model inspection

```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
seq_1_model = pd.read_csv('../data/model_outputs/3_18_Seq_Model_large_val_test_pred_df.csv')
seq_gene_1_model = pd.read_csv('../data/model_outputs/3_18_Seq_Gene_Model_large_val_test_pred_df.csv')
image_1_model = pd.read_csv('../data/model_outputs/3_18_Image_Model_large_val_test_pred_df.csv') 

groundtruth_df = pd.read_csv('../data/new_groundtruth_df_target_data_v1.csv')

seq_1_model = seq_1_model.merge(groundtruth_df, on='curve_idx', how='left')
seq_gene_1_model = seq_gene_1_model.merge(groundtruth_df, on='curve_idx', how='left')
image_1_model = image_1_model.merge(groundtruth_df, on='curve_idx', how='left')

print(image_1_model.head())
```

```python
# Calculate the AUC score

# Function to print AUC scores for a given model and dataset
def print_auc_scores(model, model_name):
    print(f'on val dataset for {model_name}')
    model_val = model[model['split_x'] == 'val']
    print(model_val.shape)
    auc_score_val = roc_auc_score(model_val['groundtruth_target'], model_val['outputs'])
    print(f"{model_name} AUC score on val: {auc_score_val}")
    auc_score_igi = roc_auc_score(model_val['groundtruth_target'], model_val['Igi_call_quant'])
    print(f"IGI AUC score on val: {auc_score_igi}")
    print()

    print(f'on test dataset for {model_name}')
    model_test = model[model['split_x'] == 'test']
    print(model_test.shape)
    auc_score_test = roc_auc_score(model_test['groundtruth_target'], model_test['outputs'])
    print(f"{model_name} AUC score on test: {auc_score_test}")
    auc_score_igi = roc_auc_score(model_test['groundtruth_target'], model_test['Igi_call_quant'])
    print(f"IGI AUC score on val: {auc_score_igi}")
    print()
    print('-----------------------------------------------')
    print()

# For seq_1_model
print_auc_scores(seq_1_model, 'seq_1')

# For seq_gene_1_model
print_auc_scores(seq_gene_1_model, 'seq_gene_1')

# For image_1_model
print_auc_scores(image_1_model, 'image_1')
```
