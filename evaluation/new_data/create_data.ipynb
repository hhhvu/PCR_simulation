### Test

```python
   
groundtruth_df = pd.read_csv(os.path.join(new_data_dir,'new_groundtruth_df.csv'))  # Load your DataFrame here
test_df = groundtruth_df[groundtruth_df.split == 'test']

test_df.loc[:,['groundtruth_target']] = 1*(test_df.groundtruth == 1)
#Define two error targets
test_df.loc[:,'Igi_call_quant'] = 1*(test_df.igi_call == 'Positive')
test_df['igi_fp'] = (test_df['Igi_call_quant'] > test_df['groundtruth_target']).astype(int)
test_df['igi_fn'] = (test_df['Igi_call_quant'] < test_df['groundtruth_target']).astype(int)


test_sample_curve_dict = {}
for i in test_df.curve_idx.unique():
    df = test_df[test_df.curve_idx == i]
    test_sample_curve_dict[i] = df['Fn'].to_list()

with open(os.path.join(new_data_dir,'new_test_curve_dict.pkl'),'wb') as handle:
    pickle.dump(test_sample_curve_dict, handle, protocol=pickle.HIGHEST_PROTOCOL)

test_df = test_df[test_df.cycle_no == 40]
test_df.to_csv(os.path.join(new_data_dir, 'test_target_data.csv'), index=False)
```

```python
join_df = pd.read_pickle('/media/ssd1/huong/PCR-huong/data/new_data/new_join_df.pkl')
retest_df = (join_df.loc[~join_df.retest_sample_id_1.isna(), ]
            .drop_duplicates())
```

### retested

```python
retest_sample = (retest_df
                 .melt(id_vars=['curve_idx','sample_barcode','sample_id','final_patient_result'], 
                       value_vars=['retest_sample_id_1','retest_sample_id_2'],
                       var_name='retest_sample_order',
                       value_name='retest_sample_barcode'
                       )
                 )

retest_sample.loc[:,'retest_cnt'] = retest_sample.groupby('sample_barcode').retest_sample_barcode.transform('nunique')
retest_sample_barcode = (retest_sample[['sample_barcode','sample_id','final_patient_result',
                                          'retest_sample_order','retest_sample_barcode','retest_cnt']].drop_duplicates()
                         .merge(join_df[['sample_barcode','sample_id','final_patient_result']].drop_duplicates(),#.rename(columns={'sample_barcode':'retested_sample_barcode',
                                                                                                      #    'sample_id':'retested_sample_id'}),
                                   how='left',
                                   left_on='retest_sample_barcode',
                                   right_on='sample_barcode',
                                   suffixes=['_orig','_retested'])
)
retest_sample_barcode.loc[:,'retested_sample_found'] = retest_sample_barcode.groupby('sample_barcode_orig').sample_barcode_retested.transform('nunique')
```

```python
print('Number of samples with retest in total:', retest_sample_barcode.sample_barcode_orig.nunique())
print('Number of samples with 1 retests:', retest_sample_barcode[retest_sample_barcode.retest_cnt == 1].sample_barcode_orig.nunique())
print('Number of samples with 2 retests:', retest_sample_barcode[retest_sample_barcode.retest_cnt == 2].sample_barcode_orig.nunique())
print('Number of samples that have all 1 retested samples found:', retest_sample_barcode[(retest_sample_barcode.retest_cnt == 1) & (retest_sample_barcode.retest_cnt == retest_sample_barcode.retested_sample_found)].sample_barcode_orig.nunique())
print('Number of samples that have all 2 retested samples found:', retest_sample_barcode[(retest_sample_barcode.retest_cnt == 2) & (retest_sample_barcode.retest_cnt == retest_sample_barcode.retested_sample_found)].sample_barcode_orig.nunique())
print('Number of samples that have less retested found:', retest_sample_barcode[retest_sample_barcode.retest_cnt > retest_sample_barcode.retested_sample_found].sample_barcode_orig.nunique())
print('Number of samples that could not find any retested samples:', retest_sample_barcode[retest_sample_barcode.retested_sample_found == 0].sample_barcode_orig.nunique())

print('\n\nNumber of retested samples found:', retest_sample_barcode.sample_barcode_retested.nunique())
```
